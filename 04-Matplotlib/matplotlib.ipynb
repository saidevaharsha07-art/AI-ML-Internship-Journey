```python
import matplotlib.pyplot as plt
import numpy as np
```

## We can draw different graphs using mathplotlib.

1.   Plot Graph
2.   Scatter Graph

3.   Bar Graphs
4.   Pie Charts





```python
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y)
plt.show()
```

**Marker:**

```python
# 'o' -> Dot marker
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='o')
plt.show()
```

```python
# '*' -> Star marker
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='*')
plt.show()
```

```python
# ',' -> comma marker
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker=',')
plt.show()
```

```python
# 'X' -> X marker
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='X')
plt.show()
```

```python
# '+' -> Plus marker
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='+')
plt.show()
```

```python
# 'P' -> Thicker Plus marker
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='P')
plt.show()
```

```python
# 'S' -> square marker
# 'D' -> diamond marker
# 'H' -> Hexagone marker
# 'p' -> pentagon marker
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='h')
plt.show()
```

**Marker Size:**

```python
# ms -> Marker Size
# mec -> Marker Edge Color
# mfc -> marker Face Color
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='h',ms=20,mec='red',mfc='g')
plt.show()
```

**Line Style:**

```python
# 1. Solid
# 2. Dotted
# 3. Dashed
# 4. Dashdot
# 5. None
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='h',linestyle='dotted')
plt.show()
```

**Marker Color:**

```python
# lw -> line Width
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y,marker='h',color="g",lw=15.5)
plt.show()
```

**Lables:**

```python
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y)
plt.title("Student Info")
plt.xlabel("Student Number")
plt.ylabel("Student Marks")
plt.show()
```

```python
#Lable Colors:
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y)
plt.title("Student Info",color='green')
plt.xlabel("Student Number",color='red')
plt.ylabel("Student Marks",color='red')
plt.show()
```

```python
# Location changing of labels and title.
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y)
plt.title("Student Info",loc="left")
plt.xlabel("Student Number",loc="right")
plt.ylabel("Student Marks",loc="top")
plt.show()
```

**Font Family:**

```python
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y)
font1={
    'family':'serif',
    'color':'blue',
    'size':20
}
font2={
    'family':'serif',
    'color':'red',
    'size':15
}
plt.title("Student Info",fontdict = font2)
plt.xlabel("Student Number",fontdict = font1)
plt.ylabel("Student Marks",fontdict = font1)
plt.show()
```

**Grids:**

```python
x = np.array([1,2,3,4,5,6])
y = np.array([2,14,16,8,10,12])
plt.plot(x,y)
plt.title("Student Info")
plt.xlabel("Student Number")
plt.ylabel("Student Marks")
plt.grid(axis = 'x')
plt.show()
```

##**Histograms:**
###A Histogram is a graph used to show the distribution (spread) of numerical data.

It tells:

* how values are grouped
* which range has more data
* frequency of values

```python
marks = [40,45,50,55,60,65,70,75]
plt.hist(marks)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
#bins
#Controls number of groups.
plt.hist(marks, bins=12)
plt.show()
```

```python
#color
#Changes histogram color.
plt.hist(marks, color = "green")
plt.show()
```

```python
#edgecolor
#Adds border color.
plt.hist(marks, edgecolor="black")
plt.show()
```

```python
pushups = [2,5,7,12,15,18,20,22,24,25,27,28,30,32,35,38,42]
plt.hist(pushups, bins=[0,10,20,30,40,50], edgecolor="black")
plt.title("Push-Ups")
plt.xlabel("Number of Push-Ups")
plt.ylabel("Number of Students")
plt.show()
#What this means
#bins=[0,10,20,30,40,50]
#Creates ranges:
#0-9
#10-19
#20-29
#30-39
#40-49
```

##**Subplot:**
###Subplot allows multiple graphs in one figure/window.

Instead of opening many windows:

* All graphs appear together.

```python
x = [1,2,3]
y1 = [10,20,30]
y2 = [5,15,25]
plt.subplot(2,1,1)
plt.plot(x,y1)
plt.subplot(2,1,2)
plt.plot(x,y2)
plt.show()
```

```python
x = [1,2,3]
y1 = [10,20,30]
y2 = [5,15,25]
plt.subplot(2,1,1)
plt.plot(x,y1)
plt.title("Student Info",color='green')
plt.xlabel("Student Number",color='red')
plt.ylabel("Student Marks",color='red')
plt.subplot(2,1,2)
plt.plot(x,y2)
plt.title("Student Info",color='green')
plt.xlabel("Student Number",color='red')
plt.ylabel("Student Marks",color='red')
plt.show()
```

```python
# Default :plt.tight_layout()
x = [1,2,3]
y1 = [10,20,30]
y2 = [5,15,25]
plt.subplot(2,1,1)
plt.plot(x,y1)
plt.title("Student Info",color='green')
plt.xlabel("Student Number",color='red')
plt.ylabel("Student Marks",color='red')
plt.subplot(2,1,2)
plt.tight_layout()
plt.plot(x,y2)
plt.title("Student Info",color='green')
plt.xlabel("Student Number",color='red')
plt.ylabel("Student Marks",color='red')
plt.show()
```

```python
# User defined Space.
x = [1,2,3]
y1 = [10,20,30]
y2 = [5,15,25]
plt.subplot(2,1,1)
plt.plot(x,y1)
plt.title("Student Info",color='green')
plt.xlabel("Student Number",color='red')
plt.ylabel("Student Marks",color='red')
plt.subplot(2,1,2)
#plt.tight_layout()
plt.subplots_adjust(hspace=0.5)
plt.plot(x,y2)
plt.title("Student Info",color='green')
plt.xlabel("Student Number",color='red')
plt.ylabel("Student Marks",color='red')
plt.show()
```
